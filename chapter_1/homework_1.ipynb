```python
# Q2. Records count
url = "https://raw.githubusercontent.com/DataTalksClub/machine-learning-zoomcamp/main/cohorts/2026/data/car_fuel_efficiency_2026.csv"
df = pd.read_csv(url)
len(df)
```

```python
# Q3. Fuel types
df["fuel_type"].nunique()
```

```python
# Q4. Missing values
df.isna().sum()
```

```python
# Q5. Max fuel efficiency
df["fuel_efficiency_mpg"].max()
```

```python
# 1. Mediana original
median_before = df["horsepower"].median()

# 2. mode 
mode = df["horsepower"].mode()[0]

# Fill
df["horsepower"] = df["horsepower"].fillna(mode)

# median_after
median_after = df["horsepower"].median()

print(f"Median before fill: {median_before}")
print(f"Median after fill: {median_after}")
```

```python
# Q7. Sum of weights
cond1=df["origin"]=="Asia"
cols=["vehicle_weight" , "model_year"]
X=df.loc[cond1,cols].head(7).values

XTX = X.T @ X
XTX_inv = np.linalg.inv(XTX)
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])

w = XTX_inv @ X.T @ y
w.sum()
```
